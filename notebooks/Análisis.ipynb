```python
import mne
from pathlib import Path
import os
import gc

BASE_PATH = Path.cwd() / "../data/cap-sleep-database"

raw = mne.io.read_raw_edf((BASE_PATH / "brux1.edf"), preload=True)
# Convertir a DataFrame
df = raw.to_data_frame()

# ---------------------------
# 1. Dimensiones
# ---------------------------
print("\n📊 DIMENSIONES")
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

# ---------------------------
# 2. Primeras filas
# ---------------------------
print("\n🔍 PRIMERAS 5 FILAS")
print(df.head())

# ---------------------------
# 3. Tipos de datos
# ---------------------------
print("\n🧠 TIPOS DE DATOS")
print(df.dtypes)

# ---------------------------
# 4. Valores nulos
# ---------------------------
print("\n🧹 VALORES NULOS")
print(df.isnull().sum())

# ---------------------------
# 5. Resumen estadístico
# ---------------------------
print("\n📈 RESUMEN ESTADÍSTICO")
print(df.describe())

# ---------------------------
# 6. Estadísticas extra útiles
# ---------------------------
print("\n📊 ESTADÍSTICAS ADICIONALES")

for col in df.columns:
    print(f"\n{col}")
    print("min:", df[col].min())
    print("max:", df[col].max())
    print("mean:", df[col].mean())
    print("std:", df[col].std())

# Liberar memoria
del df
del raw
gc.collect()
```

```python
from collections import Counter

edf_files = list(BASE_PATH.glob("*.edf"))

channel_counter = Counter()

total_files = len(edf_files)

# ---------------------------
# recorrer archivos
# ---------------------------
for file in edf_files:
    raw = mne.io.read_raw_edf(file, preload=False)

    # normalizar nombres
    channels = set([ch.strip().upper() for ch in raw.ch_names])

    # contar presencia (no duplicados)
    channel_counter.update(channels)

# ---------------------------
# calcular porcentajes
# ---------------------------
print(f"Total de pacientes: {total_files}\n")

for ch, count in channel_counter.most_common():
    percentage = (count / total_files) * 100
    print(f"{ch:15} -> {count:3} pacientes ({percentage:.2f}%)")
```
